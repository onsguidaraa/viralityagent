```python
def is_valid_tiktok_url(url):
    return "/video/" in url and "404" not in url

tiktok_urls = [u for u in tiktok_urls if is_valid_tiktok_url(u)]
```

```python
import glob
import json
import pandas as pd

rows = []

for fp in glob.glob(f"{TT_DIR}/*.info.json"):
    with open(fp, "r", encoding="utf-8") as f:
        data = json.load(f)

    rows.append({
        "platform": "tiktok",
        "video_id": data.get("id"),
        "url": data.get("webpage_url"),
        "creator_username": data.get("uploader") or data.get("channel"),
        "caption_or_title": data.get("description") or data.get("title"),
        "publish_date": data.get("upload_date"),
        "views": data.get("view_count"),
        "likes": data.get("like_count"),
        "comments": data.get("comment_count"),
        "duration_seconds": data.get("duration"),
        "local_video_path": f"{TT_DIR}/{data.get('id')}.{data.get('ext')}" if data.get("id") and data.get("ext") else None,
        "source_query": "manual_link_list"
    })

df_tt = pd.DataFrame(rows)
csv_path = f"{META_DIR}/tiktok_metadata0.csv"
df_tt.to_csv(csv_path, index=False)

print("Saved CSV to:", csv_path)
df_tt.head(10)
```
